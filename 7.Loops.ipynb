# ***-><u> While loop</u>***

Ans - A while loop executes a block of statements as long as the condition is True.
- The loop stops when the condition becomes False.
- [Infinite Loop] = An infinite loop occurs when the condition always remains True.
- stop infinite loop by press [ctrl+c]


```python
# Print Hellow World 5 times
count = 1   # iterator
while (count<=5):
    print("Hellow World")
    count += 1
```

### **Q - Break Statement**

Ans - break is used to immediately terminate the loop.
- When break is executed, the loop stops and control moves to the next statement after the loop.

```python
# breal the loop when find the multiple of 7

i = 1
while (i<=10):
    if (i%7 == 0):
        break
    print(i)
    i += 1

print("Outside of the loops...")
```

---

### **Q - Continue Statement**

Ans - continue is used to skip the current iteration.
- It moves the control to the next iteration of the loop.

```python
i = 1
while (i<=10):
    if (i%7 == 0):
        i += 1
        continue
    print(i)
    i += 1
    

print("Outside of the loops...")
```

Q - Count Digits: 
- Take an integer input from the user and count the total number of digits it 
contains using a while loop. 

```python
num = int(input("Enter your number: "))
num = abs(num) # for positive no
print(num)
count = 0
if num == 0:
    count = 1 # for 0

while (num > 0):
    count += 1
    num = num // 10
print(count)
```

Q -  Skip Multiples of 5: 
- Write a program that prints numbers from 1 to 50. If a number is a 
multiple of 5, use the continue statement to skip printing it. 

```python
num = 1
while (num<= 50):
    if (num % 5 == 0):
        num += 1
        continue
    print(num)
    num += 1
```

Q - Early Exit on Negative Number: 
- Continuously accept numbers from the user in an 
infinite while True loop. Accumulate the sum of these numbers. If the user enters a 
negative number, use break to terminate the loop and display the final sum. 

```python
total = 0

while (True):
    num = int(input("Write your no :"))
    print(num)
    if num <0:
        break

    total += num
    
print("Sum Of all positive no ",total)
```

Q - Multiplication Table (Filtered): 
- Generate the multiplication table for a given number 
$N$ from 1 to 10. Use continue to skip printing results that are even numbers. 

```python
num = int(input())
i = 1
while (i <= 10):
    result = ( i * num)
    if ( result % 2 == 0):
        i += 1
        continue
    print(num,"X",i,"=",result)
    i += 1
```

Q -  Reverse an Integer: 
- Take an integer input from the user and reverse it mathematically 
using a while loop (e.g., inputting 1234 results in 4321). 

```python
num = int(input("Enter your no:"))
print("Original No : ", num)
reverse = 0
while (num >0):
    digit = num % 10
    reverse = reverse *10+ digit

    num = num // 10
print("Reverse no : ",reverse)
```

Q - Binary to Decimal: 
- Take a binary number (as an integer consisting of only 0s and 1s) 
and convert it to its decimal equivalent using a while loop. 

```python
num =  int(input("write your binary no "))
print("Binary no = ",num)

decimal = 0
power = 0

while(num>0):
    digit = (num % 10 )
    decimal = (decimal + (digit)*(2** power))

    power += 1

    num = num // 10

print(decimal)


```
