## **---> Loops**
Ans - Loops is used to when we do repeat any work again and again.
- A loop is used to execute a block of code repeatedly until a specified condition is met.

### **-> Types of Loops in Python**
1. for loop
2. while loop

---

## **--> While Loop**

Ans - With the while loop we can execute a set of statements as long as a condition is true.

```python
i = 0
while ( i < 10):
    i += 1
    print(i)
```

---

## **--> For Loop**
Ans - A for loop is used when we know in advance how many times we want to repeat a block of code.
- A for loop is used for iterating over a sequence such as a list, tuple, string, or range.


## **Range function**

Ans - it is used for sequence generation
- range has 3 parameter [ start , stop , step]

- range (start , stop , step)

- range () - starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

```python
string = "Hello"

for var in string: # in is a membership operator ( it used to check presence )
    print(var)

```

```python
for i in range(5):
    print(i+1)
```

```python
word = "artificial intelligence"
count = 0
for ch in word:
    if (ch == "i"):
        count += 1
print(count)
```

```python
# find the vowel 
word = "artificial"
count = 0
for ch in word:
    if (ch == "a" or ch == "e" or ch =="i" or ch == "o" or ch == "u"):
     count += 1
print(count)
```

```python
# range Q 
for i in range(3):
    print(i)

print("="*25)

for ch in range (1,10):
    print(ch)

print("="*25)

for ch in range (50,0,-2):
    print(ch)
```

```python
# sum of n numbers 
n = int (input("Write no here"))

sum = 0
for i in range(1,n+1):
    sum += i
print(sum)
```

```python
# break statement in for loop 

fruit = ["banana", "Apple", "Cherry", "kiwi"]

for i in fruit:

    if (i == "Cherry"):
        break
    print(i)
```

```python
# continue
fruits = ["apple", "banana", "cherry"]
for x in fruits:
  if x == "banana":
    continue
  print(x)
```

```python
row = int(input())
print(row)
for i in range(row):
    spaces = " " * (row -i -1)
    stars = "*" * (2*i+1)
    print(spaces+stars)
```
