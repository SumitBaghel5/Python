# ***-><u> While loop</u>***

Ans - A while loop executes a block of statements as long as the condition is True.
- The loop stops when the condition becomes False.
- [Infinite Loop] = An infinite loop occurs when the condition always remains True.
- stop infinite loop by press [ctrl+c]


```python
# Print Hellow World 5 times
count = 1   # iterator
while (count<=5):
    print("Hellow World")
    count += 1
```

```python
# A infinite Loop

i = 1
while (True):
    print("Hi")
```

Q - Break Statement

Ans - break is used to immediately terminate the loop.
- When break is executed, the loop stops and control moves to the next statement after the loop.

```python
# breal the loop when find the multiple of 7

i = 1
while (i<=10):
    if (i%7 == 0):
        break
    print(i)
    i += 1

print("Outside of the loops...")
```

---

Q - continue Statement

Ans - continue is used to skip the current iteration.
- It moves the control to the next iteration of the loop.

```python
i = 1
while (i<=10):
    if (i%7 == 0):
        i += 1
        continue
    print(i)
    i += 1
    

print("Outside of the loops...")
```
