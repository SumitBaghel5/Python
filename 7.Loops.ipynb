## **---> Loops**
Ans - Loops is used to when we do repeat any work again and again.
- A loop is used to execute a block of code repeatedly until a specified condition is met.

### **-> Types of Loops in Python**
1. for loop
2. while loop

---

## **--> While Loop**

Ans - With the while loop we can execute a set of statements as long as a condition is true.

```python
i = 0
while ( i < 10):
    i += 1
    print(i)
```

---

## **--> For Loop**
Ans - A for loop is used when we know in advance how many times we want to repeat a block of code.
- A for loop is used for iterating over a sequence such as a list, tuple, string, or range.


```python
string = "Hello"

for var in string: # in is a membership operator ( it used to check presence )
    print(var)

```

```python
for i in range(5):
    print(i+1)
```
