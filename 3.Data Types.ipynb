### **Q - Tuple**
Ans - A tuple is a collection of different types of data.

* Elements of a tuple are separated by commas (,) and all elements are enclosed within parentheses ( ( ) ).
* You can also nest a tuple inside another tuple.
* Tuples are immutable, meaning they cannot be modified after creation.
* You Add a List inside a Tuple.

```python
# Example -1 
tuple = (23.5,86,"ram",(-56,26),("apple","banana"))
print(tuple)
```

```python
# Example - 2 ( change a Tuple- Show Error)
tuple = (23.5,86,"ram",(-56,26),("apple","banana"))
tuple(1) = 100
print(tuple)
```

```python
# Example - 3 ( list in tuple and change it these list)
tuple = (23.5,86,"ram",[-56,26],("apple","banana"))
tuple [3][0] = -526
print(tuple)
```

### **Q - Range**
Ans - The range data type represents an immutable sequence of numbers, commonly used for looping a specific number of times in for loops.
- It follows the syntax: range(start, stop, step).
- Start: The beginning of the sequence (default is 0).
- Stop: The sequence ends before reaching this number (exclusive).
- Step: The difference between each number in the sequence (default is 1).
- The range type is immutable (you cannot change the numbers inside the sequence).

```python
numbers = range(1, 10, 2)

print(type(numbers))  
print(list(numbers))  
```

### **Q - Dictionary**
Ans - A dictionary is an ordered collection of data that stores items in key:value pairs.
- Key-value pairs are enclosed within curly brackets { }.
- Each pair is separated by a comma (,), and the key is separated from its value by a colon (:).
- Keys must be unique, whereas values can be repeated.
- Dictionaries are mutable, allowing you to modify them after they are created.

```python
my_dict = {
    "name": "Ram",
    "age": 25,
    "city": "Delhi"
}


print(my_dict["name"],"\n",my_dict["age"])
```

### **Q - Set**

### **Q - Type conversion**
Ans - Type conversion is the process of converting one data type into another data type.

**--->There are two types**



### **1. Implicit Type Conversion (Automatic)**
- Conversion done automatically by Python.
- Python converts smaller data types into larger compatible data types to avoid data loss.

```python
a = 5
b = 3.0

print(a + b)
print(type(a + b))
```

### **2.Explicit Type Conversion (Type Casting)**
- Conversion done manually by the programmer using functions like int(), float(), str(), etc.


```python
a = "10"

print(int(a))
print(type(a))
print(type(int(a)))
```
