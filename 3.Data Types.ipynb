### **Q - Tuple**
Ans - A tuple is a collection of different types of data.

* Elements of a tuple are separated by commas (,) and all elements are enclosed within parentheses ( ( ) ).
* You can also nest a tuple inside another tuple.
* Tuples are immutable, meaning they cannot be modified after creation.
* You Add a List inside a Tuple.

```python
# Example -1 
tuple = (23.5,86,"ram",(-56,26),("apple","banana"))
print(tuple)
```

```python
# Example - 2 ( change a Tuple- Show Error)
tuple = (23.5,86,"ram",(-56,26),("apple","banana"))
tuple(1) = 100
print(tuple)
```

```python
# Example - 3 ( list in tuple and change it these list)
tuple = (23.5,86,"ram",[-56,26],("apple","banana"))
tuple [3][0] = -526
print(tuple)
```

### **Q - Range**
Ans - The range data type represents an immutable sequence of numbers, commonly used for looping a specific number of times in for loops.
- It follows the syntax: range(start, stop, step).
- Start: The beginning of the sequence (default is 0).
- Stop: The sequence ends before reaching this number (exclusive).
- Step: The difference between each number in the sequence (default is 1).
- The range type is immutable (you cannot change the numbers inside the sequence).

```python
numbers = range(1, 10, 2)

print(type(numbers))  
print(list(numbers))  
```
