### **Q-Data Types**
Ans - Data types are used to store different types of data in variables so Python can perform different operations.

-  Data types are categories that tell Python what kind of value a variable holds.

### **--->Types of Data Types**
Ans - Two types of data tyoes.
1. **Built-in Data Types**  -  These Data Types already available in Python.
2. **User-defined Data Types**  -  They are created by users using classes.

---

### **---> Built-in Data Types**
| Category | Data Types |
| :--- | :--- |
| **Text Type** | str |
| **Numeric Types** | int, float, complex |
| **Sequence Types** | list, tuple, range |
| **Mapping Type** | dict |
| **Set Types** | set, frozenset |
| **Boolean Type** | bool |
| **None Type** | NoneType |

### **Q - String**
Ans - String is a data type used to store text data.

**Rules** - Anything written inside quotes (" " or ' ') is considered a string.

**A string can contain**
- Alphabets = A to Z or a to z
- Numbers = 0 to 9
- Symbol  = @,#,$,& etc
- Spaces = Hi    Sumit ___


```python
name = "Hello my name is Sumit Baghel."
print(name)
```
