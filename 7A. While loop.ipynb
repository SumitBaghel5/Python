### **Q - Break Statement**

Ans - break is used to immediately terminate the loop.
- When break is executed, the loop stops and control moves to the next statement after the loop.

```python
# breal the loop when find the multiple of 7

i = 1
while (i<=10):
    if (i%7 == 0):
        break
    print(i)
    i += 1

print("Outside of the loops...")
```

---

### **Q - Continue Statement**

Ans - continue is used to skip the current iteration.
- It moves the control to the next iteration of the loop.

```python
i = 1
while (i<=10):
    if (i%7 == 0):
        i += 1
        continue
    print(i)
    i += 1
    

print("Outside of the loops...")
```

Q -  Reverse an Integer: 
- Take an integer input from the user and reverse it mathematically 
using a while loop (e.g., inputting 1234 results in 4321). 

```python
num = int(input("Enter your no:"))
print("Original No : ", num)
reverse = 0
while (num >0):
    digit = num % 10
    reverse = reverse *10+ digit

    num = num // 10
print("Reverse no : ",reverse)
```

and convert it to its decimal equivalent using a while loop. 

```python
num =  int(input("write your binary no "))
print("Binary no = ",num)

decimal = 0
power = 0

while(num>0):
    digit = (num % 10 )
    decimal = (decimal + (digit)*(2** power))

    power += 1

    num = num // 10

print(decimal)


```

Q - Password Validation (Limited Attempts):
- Create a program with a predefined password. 
- Allow the user up to 3 attempts to enter the correct password using a while 
loop. Use break upon a successful login, or notify them if they run out of attempts. 

```python
password = "admin123"   # predefined password
attempts = 3

while attempts > 0:
    user_password = input("Enter Password: ")

    if user_password == password:
        print("Login Successful!")
        break

    attempts -= 1
    print("Wrong Password!")
    print("Attempts Left:", attempts)

if attempts == 0:
    print("You have run out of attempts.")


```
