## **-> Check String**
Ans - To check whether a word, character, or substring exists in a string.
- Python uses the **in** and **not** in operators.

### **-> Using in Operator** 
= The in operator returns True if the specified text is present in the string; otherwise, it returns False.

````python
name = "Sumit Baghel"

print("Sumit" in name )

print("="*50)

print("sumit" in name)
````

### **-> Using not in Operator**
= The not in operator returns True if the specified text is not present in the string.

````python
name = "Sumit Baghel"

print("Sumit" not in name )

print("="*50)

print("sumit" not in name)
````

## **-> Slicing Strings**

Ans - String slicing means extracting (taking) a part of a string.

- It allows you to divide a string into smaller parts.
- Python uses the slice operator `[:]` for slicing strings.
- String indexing starts from **0**.
- The start index is included, but the end index is excluded.

### Syntax

string_name[start : end : step]

Example:
```python
text = "Python"

print(text[0:2:1])

````python
text = "python"
print(text[0:3:1])
````

## **->Slice From the Start**
Ans - The range will start at the first character

````python
b = "Hello, World!"
print(b[:5])
````

## **->Slice To the End**
Ans - By leaving out the end index, the range will go to the end.


````python
b = "Hello, World!"
print(b[5:])
````

## **->Negative Indexing**
Ans - Use negative indexes to start the slice from the end of the string.

````python
b = "Hello, World!"
print(b[-5:-2])
````

## **-> Modify Strings**

**Ans -** Python provides many built-in string methods to modify strings.

Some commonly used string methods are:

- Convert to Uppercase → `upper()`
- Convert to Lowercase → `lower()`
- Remove White Spaces → `strip()`
- Split a String → `split()`
- Replace a String → `replace()`

> **Note:** Strings are **immutable** in Python. String methods do not change the original string; they return a **new string**.

````python
text= " Anuj - singh , PArihar  "
print(text.upper())
print(text.lower())
print(text.strip())
print(text.split(",")) #It returns the result as a list.
print(text.replace("A","M"))
````

## **--> Format - Strings**
Ans - String formatting means adding variables or values inside a string to create a complete and meaningful sentence.

Instead of writing values manually, Python inserts them automatically.

````python
name = "Sumit"
age = 23

print("My name is " + name + " and I am " + str(age) + " years old.")
````

## **-> The most two method of formating a string**
### **1 - format()function** - The format( ) method inserts values into placeholders `{ }` in a string.
### **2 - f-strings** - f-string (Formatted String Literal) is a way to insert variables, values, or expressions directly inside a string using curly braces {}. It starts with the letter f before the string.
- It is the easiest and most recommended way to format strings in Python.

##

````python
# Ex - Format string
name = "Monika"
age = 23

print("My name is {} and I am {} years old".format(name,age))

print("="*50)
                                        #this hold 0, and this 1
print("My name is {1} and I am {0} years old".format(name,age)) # We can also used positional & named placeholders
````

````python
# Ex - f-string
name = "Monika"
age = 23

print(f"My name is {name} and I am {age} years old.")

# ex - 2
a  = int(input("1st no = "))
b = int(input("2nd no = "))

print(f"sum of {a} & {b} = {a + b}")
print(f'Average of {a} & {b} is { (a+b)/2}')
````

## **->Escape Character
Ans - An escape character is a backslash `(\)` followed by another character. It is used to represent special
 characters inside a string that cannot be typed directly or have a special meaning.

Why Do We Use Escape Characters?

We use escape characters to:

- Print quotation marks inside a string.
- Move the text to a new line.
- Add tabs (spaces).
- Print a backslash (\).


| Escape Character | Description | Example |
|------------------|-------------|---------|
| `\'` | Single quote | `'It\'s Python'` |
| `\"` | Double quote | `"He said, \"Hello\""` |
| `\\` | Backslash | `"C:\\Users\\Sumit"` |
| `\n` | New line | `"Hello\nWorld"` |
| `\t` | Tab (horizontal space) | `"Python\tJava"` |
| `\b` | Backspace | `"Hell\bo"` |
| `\r` | Carriage return | `"Hello\rHi"` |

````python
print("my name\'s sumit baghel")
print("my name\"s sumit baghel")
print("my name\rsumit baghel")
````
