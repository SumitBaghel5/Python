### **1. Movie Ticket Price Calculator**

A console program that determines individual cinema admission costs by processing patron age, session
timing, and educational affiliation discounts.

- Variables & Data Types: Age (int), ShowTime (str e.g., "Matinee", "Evening"), and IsStudent (bool).
- Operators: Comparison operators (<=, >=) to evaluate age boundaries.
- Conditionals: if-elif-else tiers providing cheaper entry rates for children, senior citizens, or midday
matinee show times.

```python
# 1. Inputs and Data Types
name = input("Enter your Name = ")
age = int(input("Enter your Age = "))
student = input("Type Y or N = ")
show_time = input("Type Morning, Mid_Day, Evening = ")
ticket_price = int(input("Ticket price = "))

# 2. Age group math logic bina if-else ke
# (age < 18) True hoga to 1 banega, False hoga to 0 banega.
is_child = int(age < 18)
is_adult = int(18 <= age <= 50)
is_senior = int(age > 50)

# Alag-alag conditions ke discounts (Rupees mein)
discount_age_amt = (is_child * 0.20 * ticket_price) + (is_adult * 0.10 * ticket_price) + (is_senior * 0.30 * ticket_price)

# 3. Show Time aur Student ke liye Look-up Dictionaries (Data Types)
# Agar user "Morning" likhega to dictionary use 0.30 (30%) return karegii.
show_discounts = {"Morning": 0.30, "Mid_Day": 0.20, "Evening": 0.10}
student_discounts = {"Y": 0.20, "N": 0.10}

# .get() ka use karke value nikalna. Agar user galat type karega to 0 discount milega.
discount_show_amt = show_discounts.get(show_time, 0.0) * ticket_price
discount_student_amt = student_discounts.get(student, 0.0) * ticket_price

# 4. Total calculation (Sirf Operators ka use karke)
total_discount_saved = discount_age_amt + discount_show_amt + discount_student_amt
calculated_price = ticket_price - total_discount_saved

# 5. Output Formatting (Bina Tuple ke, taaki single line m print ho)
final_price = f"Final Price of your movie ticket = Rs. {calculated_price}"

# 6. Printing 
print("Name = ", name)
print("Age = ", age)
print("Student = ", student)
print("Show Time = ", show_time)
print("Ticket Price = ", ticket_price)
print("Final Price = ", final_price)
```

```python
# Temp converter C to F or F to C

n = float(input("write in no "))
temp = input("temp in degree ")
print("temp = ",n,temp)
# print("original degree = ",temp)

if (temp == "C"):
    result = (n*1.8)+32
    print(f"Temp in Fahrenheit = {result:.2f}","F")
elif (temp == "F"):
    result = (n-32)*5/9
    print(f"Temp in Fahrenheit = {result:.2f}","C")
else:
    print("Invalid input. Please enter 'C' or 'F'.")
```

## **9. Shipping Cost & Delivery Time Estimator**

An e-commerce tool estimating dispatch handling charges and transit fulfillment timelines using payload
weight metrics and distance zones.
- Variables & Data Types: PackageWeight (float) and DestinationZone (str: "Local", "National", "International").
- Operators: Mathematical tracking calculating base flat-rates compounded by precise weight surcharges.
- Conditionals: Branch structures determining final shipping profiles and ETA days depending on the selected
destination zone.

```python
# 1. Take inputs from the user
product_weight = float(input("Enter your product weight in kg: "))
destination_zone = input("Enter destination zone (Local, National, International): ").strip().capitalize()

total_shipping_cost = 0

# 2. Tiered Shipping Cost Calculation
if product_weight <= 50:
    # Everything falls into the base tier rate
    if destination_zone == "Local":
        total_shipping_cost = product_weight * 2000
    elif destination_zone == "National":
        total_shipping_cost = product_weight * 3000
    elif destination_zone == "International":
        total_shipping_cost = product_weight * 4000
    else:
        print("Invalid zone entered.")

else:
    # Weight is over 50 kg. We split the calculation.
    extra_weight = product_weight - 50
    
    if destination_zone == "Local":
        # 50 kg at the base rate + extra weight at the premium rate
        total_shipping_cost = (50 * 2000) + (extra_weight * 3000)
    elif destination_zone == "National":
        total_shipping_cost = (50 * 3000) + (extra_weight * 4000)
    elif destination_zone == "International":
        total_shipping_cost = (50 * 4000) + (extra_weight * 5000)
    else:
        print("Invalid zone entered.")

# 3. Display the final cost
if total_shipping_cost > 0:
    print(f"\nFinal Shipping Cost: {total_shipping_cost:,}")
```
