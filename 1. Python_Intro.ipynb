### **Q - What is Translator or Compiler ?**

Ans - A Translator or Compiler is a software that converts programming language into machine language (0 and 1) so that the computer can understand and execute the program.

Computers only understand binary language (0 and 1).  
The translator helps convert programming language into 0 & 1 language so programs can run on the computer.

**Examples of Translators**
- Compiler
- Interpreter
- Assembler

### **Q - Python Indentation**

Ans - Indentation is a space begining of the code line.
- other programming languages (like Java, C++, or JavaScript) use curly braces { } to group blocks of code, Python uses indentation.
- It isn't just there to make your code look clean—it's a strict syntax requirement. If you get it wrong, your code won't run.

#### ***How it Works***
When you start a code block—like an if statement, a for loop, or a function—you end the initial line with a colon (:). Every line of code that belongs inside that block must be indented by the exact same amount of space.

```python
if 5 > 2:
    print("Five is greater than two!")  # This line is inside the if-block
    print("This runs too!")             # Same indentation = same block
    
print("This is outside the block.")     # Back to the start line = block over
```
