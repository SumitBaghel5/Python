## **--> For Loop**
Ans - A for loop is used when we know in advance how many times we want to repeat a block of code.
- A for loop is used for iterating over a sequence such as a list, tuple, string, or range.

Ans - Even strings are iterable objects, they contain a sequence of characters:

### **->The break Statement**
Ans - With the break statement we can stop the loop before it has looped through all the items:

````python
# if i dont want to take break  entity 
fruits = ["banana","Apple","cherry","kiwi","papaya"]

for x in fruits:
    if ( x == "cherry"):
        break
    print(x)
````

````python
# if i want with break entity
fruits = ["banana","Apple","cherry","kiwi","papaya"]

for x in fruits:
    print(x)
    if (x == "cherry"):
        break
    
````

---

### **->The continue Statement**
Ans - With the continue statement we can stop the current iteration of the loop, and continue with the next:

````python
fruits = ["banana","Apple","cherry","kiwi","papaya"]

for x in fruits:
    if ( x == "cherry"):
        continue
    print(x)
````

### **-> The range() Function**
Ans - ThefunctioninPythonisusedtogenerateasequenceofnumbers, typicallyforuseinloops.
- Itdoesn’tactuallycreatealistinmemory(unlesswe convert it); instead, it creates an iterator that produces numbers one by one, whichmakesitveryefficient.
- The range() function returns a sequence of numbers, starting from 0 by default, and increments by 1 (by default), and ends at a specified number.

### **Thefunctionhas3parameters**

- start(default=0)-thenumbertostartfrom
- stop-thenumbertostopbefore(notincluded)
- step(default=1)-how muchtoincrease byeachtime

````python
# 1 Argument
for i in range(5):
    print(i)
````

````python
# 2 Argument
for i in range (1,6):
    print(i)

print("="*52)

num  = int (input())
for i in range(1,num+1):
    print(i)
````

````python
# 3 Argument
num1 = int(input("Starting no"))
num2 = int(input("Write End no"))
num3 = int (input("step no"))

for i in range(num1,num2+1,num3):
    print(i)
````

## **-->Nested Loops**
Ans - A nested loop is a loop inside a loop.
- The "inner loop" will be executed one time for each iteration of the "outer loop":



``` text
You can imagine it as a matrix:

	j=1	j=2
i=1	 (1,1)	(1,2)
i=2	(2,1)	(2,2)

Execution flow:

i = 1
    j = 1 -> (1,1)
    j = 2 -> (1,2)

i = 2
    j = 1 -> (2,1)
    j = 2 -> (2,2)

If you want the output to actually look like a matrix, use end=" ":

````python
for i in range(1,3):
    for j in range(1,3):
        print(f"({i},{j})",end = " ")
    print()

print("="*52)

adj = ["red", "big", "tasty"]
fruits = ["apple", "banana", "cherry"]

for x in adj:
  for y in fruits:
    print(x, y )
  print()

````
