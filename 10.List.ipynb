## **->LIST**

Ans - Used to store multiple items in a single variable.
- One of four built-in collection data types in Python (along with Tuple, Set, and Dictionary).
- Created using square brackets [].
- Ordered: Items have a defined order that does not change change there index value ; new items are appended to the end.
- Changeable (Mutable): Elements can be modified, added, or removed after creation.
- Allow Duplicates: Can contain items with identical values since they are indexed.

```python
demo = ["ram", "kishor", "dhami", 56, 253,45.2313]

print(demo)
```

## **-->Core List Operations & Features**
Ans - len() function: Determines the total number of items in a list.

- list() constructor: Used to create a new list from an iterable.


```python
# find the length of th list
demo = [1,5,"ram","kishor", 65]
print(demo)
print(type(demo))
print(len(demo))

print("="*25)

# create a list throug second method
demo2 = list((2,5,96,85))
print(demo2)

print(type(demo2))
```

## **-> Accessing Items (Indexing & Slicing)**
- Positive Indexing: Starts from 0 for the first item.
- Negative Indexing: Starts from the end (-1 represents the last item, -2 the second last, etc.).
- Range of Indexes (Slicing): Returns a new list with items from the specified start index up to (excluding) the end index. Also includes the slice operator (:).

```python
# Access element of list by there index value 

demo = ["ram", "monika", 25,85,36]

print(demo)

# if i want only 2 value from the list 

print(demo[1])

print(demo[-2]) # this is neative indexing in which we take element from the back side

# Slicing of indexing

print(demo[1:3])
```
