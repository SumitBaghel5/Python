## **-> Accessing Items (Indexing & Slicing)**
- Positive Indexing: Starts from 0 for the first item.
- Negative Indexing: Starts from the end (-1 represents the last item, -2 the second last, etc.).
- Range of Indexes (Slicing): Returns a new list with items from the specified start index up to (excluding) the end index. Also includes the slice operator (:).

```python
# Access element of list by there index value 

demo = ["ram", "monika", 25,85,36]

print(demo)

# if i want only 2 value from the list 

print(demo[1])

print(demo[-2]) # this is neative indexing in which we take element from the back side

# Slicing of indexing

print(demo[1:3])
```

## **--> List method**
Ans - Python list methods are built-in functions attached to list objects that allow you to easily modify, update, search, or reorganize the elements inside a list without having to write manual loops or custom logic.


| Method | Use |
| :--- | :--- |
| `append()` | Adds a single element to the end of the list. |
| `extend()` | Adds multiple elements (from another list or iterable) to the end of the list. |
| `insert()` | Adds an element at a specific index position in the list. |
| `remove()` | Removes the first occurrence of a specified value from the list. |
| `pop()` | Removes an element at a specific index (or the last element if no index is specified) and returns it. |
| `clear()` | Removes all elements from the list, leaving it empty. |
| `index()` | Returns the index of the first occurrence of a given value. |
| `count()` | Returns the number of times a specific value appears in the list. |
| `sort()` | Arranges the list elements in ascending or descending order. |
| `reverse()` | Reverses the order of the elements in the list. |
| `copy()` | Returns a shallow copy of the list. |

```python
# Append ()
demo_list = ["apple", "banana", "cherry"]

demo_list.append("kam")  # [ Note - You append only one item at a time]
demo_list.append("yam")
#demo_list.append("koyal","jam") # its shows error 

print(demo_list) 
```

```python
# Extend()
# - extending a list means appending all the elements from an iterable (such as another list, tuple, set, or string)
#  to the end of the original list, increasing its overall length.
demo_list = list(("apple","banana","cherry"))

# another list
demo2_list = ["mango","pineapple","papaya"]

# Extend first list
demo_list.extend(demo2_list)

print(demo_list)

# Extend second list
demo2_list.extend(demo_list)

print(demo2_list)
```

```python
# Insert () [ #Note - in insert  we insert only value at a place of index not replace previous value , previouse value is extend to next index value]
demo_list = ["apple","maths","sciencs","physics","chemistry"]

demo_list.insert(0,"hindi")

print(demo_list)

# insert a list into a list
demo_list.insert(1,["hindi","english","gk","sanskrit"])
print(demo_list )
```

```python
# Remove()     [Note - The remove() method in Python removes an element by searching for its specific value, not its index.]
demo_list = ["apple","banana","cherry"]
demo_list.remove("apple")  #  [ Note - We reove only one element of list at a time]

print(demo_list)
```

```python
# Pop ()
# The pop() method removes the specified index.
demo_list = ["maths","english","hindi","sanskrit"]
demo_list.pop(1)  # [ Note - Only one element pop at a time]
print(demo_list)

# Remove the last item:
demo_list.pop()
print(demo_list)
```

```python
# Clear()  # its remove all element of the list 
# its make a list into empty list
demo_list = ["maths","english","hindi","sanskrit"]

demo_list.clear()

print(demo_list)
```

```python
# index ()
colors = ["red", "green", "blue", "green", "yellow"]

# Find the index of "blue"
print(colors.index("blue"))  # Output: 2

# Find the index of "green" (returns the first occurrence)
print(colors.index("green")) # Output: 1

# q - 

numbers = [10, 20, 30, 40, 20, 50]

# Search for 20 starting from index 2 onwards
print(numbers.index(20, 2))  # Output: 4 (skips the 20 at index 1)

# Search for 20 within a specific slice (index 1 to 4)
print(numbers.index(20, 1, 4)) # Output: 1
```

```python
# Count()
demo_list = [2,5,8,2,1,5,6,2,5,2,3,4,5,6,4,5,2,45,85,5]

print(demo_list.count(2))   # count 2 how many time
```

```python
# sort()
# arrange the elements of a list in a specific order

# default sort [Ascending order]
numbers = [42, 10, 3, 55, 21]
numbers.sort()
print(numbers)  

words = ["banana", "apple", "cherry", "date"]
words.sort(reverse = False)
print(words)    

# Sorting in Descending Order
numbers.sort(reverse = True)
print(numbers)

words.sort(reverse = True)
print(words) 


# Custom Sorting by key parameter
fruits = ["apple", "fig", "banana", "strawberry"]

# Sort by length of the string
fruits.sort(key=len)
print(fruits) 

```

```python
# reverse()
# Create a sample list
numbers = [10, 20, 30, 40, 50]

print("Original list:", numbers)

# Reverse the list in-place
numbers.reverse()

print("List after reverse():", numbers)
```

```python
# copy() - [Returns a new list that is a duplicate of the original list.]
# Create an original list
original_list = ["apple", "banana", "cherry"]

# Create a copy of the list
copied_list = original_list.copy()

# Modify the copied list
copied_list.append("date")

print("Original list:", original_list)
print("Copied list:", copied_list)
```

## **--> Loop in List**
Ans - We use the classical For loop to traverse the entire list element - wise

```python
thislist = ["apple", "banana", "cherry"]
for x in thislist:
  print(x)
```

```python
# Linear Search 
# A linear search checks each element one by one to find a target(x).
numbers = [5,7,12,3,18,9]
x = 18
idx = 0
for num in numbers:
    if num == x:
        print(f"{x} found at index={idx}")
        break
    idx += 1
```
