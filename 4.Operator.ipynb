### **Q - Operators and Operands**
Ans - Operators are symbols that perform operations on operands.
Ex - a + b
- += Operator (performs operation)
- a,b = Operands (values on which operation is performed)

## **---> <u>Python has different types of operators.<u/>**

### **1.Arithmetic Operators**
- Arithmetic operators are used to perform mathematical operations.

|Operator| Meaning | Example|
|:---|:---|:---|
|+ | Addition| 5+2 = 7|
|- | Subtraction| 5-2 = 3|
|* | Multiplication| 5*2 = 10|
| /	| Division|	5/2 = 2.5|
| ** | Exponent (Power)| 5**2 = 25|
| // | Floor Division | 5//2 = 2|
| % | Modulus (Remainder) | 5%2 = 1|


```python
a = 25
b = 69
print("Add = ",a+b)
print("Sub = ",a-b)
print("mult = ", a*b)
print("divi=",a/b)
print("power=",a**b)
print("floor div=",a//b)
print("mod=",a%b)
```

### **Q - Relational Operators**
Ans - Relational operators are used to compare two values.
- They compare values and return True or False.( they give result in boolean)
- They are also called Comparison Operators.

| Operator | Meaning | Example |
|:---|:---|:---|
| == | Equal to | 5==5 → True |
| != | Not Equal to| 5!=3 → True |
| > | Greater than | 5>3 → True |
| <	| Less than	| 3<5 → True |
| >= | Greater than or Equal to | 5>=5 → True |
| <= | Less than or Equal to | 3<=5 → True |

```python
a = 56 
b = 86
print("a = b =",a==b)
print("a!=b = ", a!=b)
print("a>b =",a>b)
print("a<b =",a<b)
print("a>=b =", a>=b)
print("a<=b =", a<=b)
```

### **Q - Assignment Operators**

Ans - Assignment operators are used to assign values to variables.

- They can also be used to update existing variable values.

Ex = a = 5( 5 is assign in a)

a = a+3 (means - a = 5 and add 3 more then a = 8 ---------  We write it a += 3)

|Operator|	Meaning|	Example|
|:---|:---|:---|
|=	|Assign value	|a = 10|
|+=|	Add and assign	|a += 5|
|-=|	Subtract and assign|	a -= 5|
|*=|	Multiply and assign	|a *= 5|
|/=|	Divide and assign	|a /= 5|
|//=	|Floor divide and assign	|a //= 5|
|%=	|Modulus and assign	|a %= 5|
|**=|	Power and assign|	a **= 2|

```python
a = 16

print("Initial value =", a)

a += 13
print("After += 13 =", a)

a -= 13
print("After -= 13 =", a)

a *= 13
print("After *= 13 =", a)

a /= 13
print("After /= 13 =", a)

a //= 13
print("After //= 13 =", a)

a %= 13
print("After %= 13 =", a)

a **= 13
print("After **= 13 =", a)
```

### **Q - Logical Operators**

Ans - Logical operators are used to combine multiple Boolean expressions or conditions.

- They return True or False
- Mainly used with conditions and decision making.

### **Types of Logical Operators**

|Val-1|Val-2|and|
|:---|:---|:---|
|T|T|T|
|T|F|F|
|F|T|F|
|F|F|F|

**2. Or** - Returns True if at least one condition is True.

|Val-1|Val-2|or|
|:---|:---|:---|
|T|T|T|
|T|F|T|
|F|T|T|
|F|F|F|

**3. not** - Reverses the result.

|Val-1|not|
|:---|:---|
|T|F|
|F|T|

```python
x = 12
y = 10
z = 10
print(x>y and x<y)
print(x>y or x<y)
print(not(x>y) )
```

### **Q - Operator Precendence**
Ans - Operators have a priority.
- it  means there is a specific order in which operations are performed when we have multiple operators appearing in the same expression.

### **The order is as follows:**
- () - Parentheses (highest priority)

- ** - Exponent

- +x, -x, ~x - Unary operators

-  /, //, % - Multiplication, division, floor di, modulus

- +, - - Addition, subtraction

- <<, >> - Bitwise shifts

- & - Bitwise AND

- ^ - Bitwise XOR

- | - Bitwise OR

- Comparison operators - <, <=, >, >=, !=, ==

- not - Logical NOT

- and - Logical AND

- or - Logical OR

- Assignment operators - =, +=, -= ...
