### **->Function with Return Statement.**

Ans -  
- A return statement sends a result back outside the function so that we can use it for other tasks in our code.
- The return statement acts as the exit point of a function. 
- The moment Python executes a return line, it exits the function immediately, and any code written after it inside the function will not run.

````python
# Function With Return Statement 

def add_with_return(a, b):
    result = a + b
    return result  
# Calling the function
total = add_with_return(5, 3) # Allways call function we need to store in some variable


print("Saved value:", total)
print("Double the total:", total * 2) # perform more operation on the result


````

### **<u>Function Argument</u>**
Ans - Arguments are the actual values passed to a function when it is called.
- They provide the data that the function's parameters receive and use.
- A function can accept one, multiple, or no arguments.

## **--> Types of Argument**
- **Positional Arguments** – Values are passed in the same order as the parameters.
- **Keyword Arguments** – Values are passed using parameter names.
- **Default Arguments** – Parameters have default values if no argument is provided.
- **Variable-Length Arguments** <u>(*args and **kwargs)</u> – Allow a function to accept multiple arguments.

````python
# 1. Positional Argumentsn / simple example  ( Simple example is a positional Arguments)
def name_1(first_name,last_name):
    print(first_name+" "+last_name)

name_1("ram","pratap")
name_1("pratap","ram")

print("=="*15)

#2. Keyword Arguments
def name_2(first_name,last_name):
    print(first_name+" "+last_name)

name_2(last_name= "Sharma",first_name ="Monika")

print("=="*15)

# 3. Default Arguments
def name_3(first_name,last_name = "Baghel"):
    print(first_name+" "+last_name)

name_3("monika")
name_3("Monika","Sharma")

print("=="*15)

# 4. Variable length Arguments

def nume_4(*num):
    print(num)
    print(type(num))

nume_4(1,2,3,5,6,4,9,8,7)
````

## **---> Variable Length Arguments**
Ans - there are two type--

### A - *args

### B- **kwargs
--> <u>*args</u> = *args is used when we want to pass multiple positional arguments into a function.
- The args parameter stores all the passed arguments in a tuple.
- Therefore, the output type of args is tuple. 


````python
def demo(*args):
    print(args)
    print(type(args))
    sum = 0
    for i in args:
        sum += i
    print(sum)

demo(1,2,3,41,56,6,1,5)
````

--> <u>**kwargs</u> = **kwargs is used when we want to pass multiple keyword arguments into a function.
- The kwargs parameter stores all the passed keyword arguments in a dictionary.
- Therefore, the output type of kwargs is dictionary.

````python
def demo(**kwargs):
    print(kwargs)
    print(type(kwargs))
    for key, value in kwargs.items():
        print(key,value)

demo(name = "sumit",age = "23",class_ = "4th")
````

## **->Decorators**
Ans - A decorator is a function that adds extra functionality to another function without changing its original code.
- A decorator is a function that takes another function as input, adds some extra functionality, and returns the modified function.

Example = 

🎁 Gift = Original function

🎀 Wrapper = Decorator

The gift stays the same, but the wrapper adds something extra.

````python
# in which we write manually 
def greet():
    print("Hello World")

print("Start function") # We add the start and end functions inside the greet function manually for a better approach.
greet()
print("End Function")
````

````python
# Decorator Function
def decorators(func):          # func stores the original function

    def greet():               # Wrapper (sub-function)
        print("Hi, This is starting")

        func()                 # Call the original function

        print("Thank you, This is end")

    return greet               # Return the wrapper function


@decorators                    # Apply the decorator
def name():
    print("Today is a very good day for me.")
    print("Now I am doing something interesting.")

name()
````

````python
#Decorator with Arguments
def decorator(func):
    def wrapper(name):
        print("Start Now")
        func(name)
        print("Stop Now") 

    return wrapper

@decorator
def dairy(name):
    print("Hello",name)

dairy("RajaRam")
````

## **-> Lambda Function**
Ans - Lambda  functions  are  short,  one-liner  functions  that  are  used  to  perform  simple tasks. 
- These are created using the **lambda** keyword. These functions are anonymous & do not have a name like regular functions defined with def.
- Can contain only one expression.
- Cannot contain multiple statements such as loops or multiple return statements.
- Not suitable for complex logic.

## **Syntax ---**
lambda arguments :  expression 
- arguments -> 1 or more parameters 
- express ion -> single expression whose result is automatically returned

````python
# Normal Function

def add(a,b):
    return a+b
print(add(10,56))

print("=="*23)

# lambda function

add = lambda a,b:a+b
print(f"sum of a+b is {add(10,6)}")
````
