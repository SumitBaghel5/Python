### **Q - Conditional Statement**

Ans - A Conditional Statement is used to make decisions in a program.
- It checks whether a condition is True or False and executes code accordingly.

--> Real-Life Example
1. If it is raining → Take an umbrella.
2. If marks are greater than 40 → Pass.
3. If age is 18 or above → Eligible to vote.

### -- > Python uses three main keywords for conditional logic: if, elif, and else.

---

### **Q - if Statement**
Ans - The **if** statement executes code only when the condition is True.

Syntax -- 

if condition:

         statement

````python
age = int(input("Enter your age - "))
if age >=18 :
    print("You are Eligible for vote")
````

````python
num = int(input("Enter your No - "))
if num > 0:
    print("Your no is positive")
````

### **Q - if - else Statement**
Ans - The if - else statement is two different paths of execution based on whether a condition is True or False.
- If the condition evaluates to True, Python executes the code inside the if block.
- If the condition evaluates to False, Python skips the if block and executes the code inside the else block.
- else condition run when the 1st condition (if) is not satified 
- Only one block (if or else) executes at a time.

Syntax --

if condition:

          # code for True condition

else:

           # code for False condition

```text
Step :                                               Check the condition

                                                            |
                                                            |
                                                            v
                                                         Condition
                                                        /         \
                                                     True         False
                                                      |             |
                                                      v             v
     
                                                   if block       else block
                                                   executes        executes

````python
age = int(input("Enter your age = "))

if (age >= 18):
    print("You are eligible for vote")
else:
    print("You are not eligible for vote")
````

````python
num = int(input("Enter your desire no here = "))

if (num >= 0):
    print("Positive No")
else :
    print("Negative No")
````

### **Q - if - elif - else Statement.**
Ans - The if-elif-else statement is a conditional statement used when there are multiple possible outcomes. 
- It allows Python to check several conditions one by one and execute the block of code associated with the first condition that evaluates to True
- if checks the first condition.
- elif (else if) checks additional conditions if the previous conditions are False.
- else executes when none of the conditions are True.

Syntax -- >
```text
if condition1:
    # code block 1

elif condition2:
    # code block 2

elif condition3:
    # code block 3

else:
    # code block 4

Step :       
```text
              Check conditions
                    |
                    v
              Condition 1 (if)
                 /      \
             True        False
              |            |
              v            v
          if block    Condition 2 (elif)
          executes       /      \
                    True        False
                     |            |
                     v            v
               elif block 1  Condition 3 (elif)
                executes       /      \
                           True        False
                            |            |
                            v            v
                   elif block 2    ... [Repeats for more elifs]
                        executes         |
                                         v
                                Condition N (last elif)
                                     /      \
                                 True        False
                                  |            |
                                  v            v
                             elif block N  else block
                              executes      executes

````python
#username = "admin" and password = "pass"
user_name = input("Enter your username = ").lower()
password = input("Enter your password = ").lower()

print("User Name = ",user_name)
print("Password = ",password)

if (user_name == "admin" and password == "pass"):
    print("Login Successful")
elif (user_name != "admin"):
    print("wrong username")
elif (password != "pass"):
    print("wrong Password")
````

### **Q - Nesting**
Ans - Nested if Statement (Nesting)
- Nesting means placing one conditional statement inside another conditional statement.
- In Python, a Nested if is an if statement written inside another if, elif, or else block.
- It is used when a second condition should be checked only after the first condition is satisfied.

Syntax -->
```text
if condition1:
    if condition2:
        # code

Q - Take the marks of 5 subjects from the user and display their individual subject grades (A, B, C, or D). The program must also calculate the total marks, overall percentage, and print the final passing division (1st, 2nd, 3rd, or Fail).

````python
maths = int(input("Enter your maths no = "))
science = int(input("enter your science no = "))
english = int(input("enter english no = "))
hindi =  int(input("Enter your hindi no = "))
sst = int(input("enter your sst no = "))

# for maths

if (maths < 20 ):
    print("maths =",maths," D-Grade")
elif (maths < 40 ):
    print("maths =",maths," C - Grade")
elif (maths < 60):
    print("maths =",maths," B - Grade")
else :
    print("maths =",maths," A - Grade")

# for science 

if (science < 20 ):
    print("science =",science," D-Grade")
elif (science < 40 ):
    print("science =",science," c-Grade")
elif (science < 60):
    print("science =",science," B-Grade")
else :
    print("science =",science," A-Grade")

# for english

if (english < 20 ):
    print("english =",english,"D-Grade")
elif (english < 40 ):
    print("english =",english,"C-Grade")
elif (english < 60):
    print("english =",english,"B-Grade")
else :
    print("english =",english,"A-Grade")

# for hindi

if (hindi < 20 ):
    print("hindi =" ,hindi, "D-Grade")
elif (hindi < 40 ):
    print("hindi =" ,hindi, "C-Grade")
elif (hindi < 60):
    print("hindi =" ,hindi, "B-Grade")
else :
    print("hindi =" ,hindi, "A-Grade")

# for sst

if (sst < 20 ):
    print("sst =" ,sst," D-Grade")
elif (sst < 40 ):
    print("sst =" ,sst," C-Grade")
elif (sst < 60):
    print("sst =" ,sst," B-Grade")
else :
    print("sst =" ,sst," A-Grade")

total_marks = (maths + science + english + hindi + sst) 
print("Total Marks = ",total_marks)
total_percentage = total_marks / 5
print("Total % = ", total_percentage,"%")

if (total_percentage <=30 ):
    print("You are failed")
elif (total_percentage <= 50):
    print("you passed with 3rd Division")
elif (total_percentage <= 70):
    print("You passed with 2nd Division ")
else:
    print("You Passed with 1st division")




````

Q - Calculate bonus: 
- Salary > 50,000 → Bonus = 20% 
- Otherwise 10%

````python
salary = float(input("Enter your salary : "))
print("Salary = ", salary)

if (salary > 50000):
    bonus_percentage = 20
else :
    bonus_percentage = 10

bonus_amount = (bonus_percentage*salary)/100
total_salary = (bonus_amount + salary)

print(" Bonus Amount is - ",bonus_amount)
print("your total salary after bonus is - ",total_salary)
````
