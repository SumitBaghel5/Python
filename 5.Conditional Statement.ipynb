### **Q - Conditional Statement**

Ans - A Conditional Statement is used to make decisions in a program.
- It checks whether a condition is True or False and executes code accordingly.

--> Real-Life Example
1. If it is raining → Take an umbrella.
2. If marks are greater than 40 → Pass.
3. If age is 18 or above → Eligible to vote.

### -- > Python uses three main keywords for conditional logic: if, elif, and else.

---

### **Q - if Statement**
Ans - The **if** statement executes code only when the condition is True.

Syntax -- 

if condition:

         statement

```python
age = int(input("Enter your age - "))
if age >=18 :
    print("You are Eligible for vote")
```

```python
num = int(input("Enter your No - "))
if num > 0:
    print("Your no is positive")
```

### **Q - if - else Statement**
Ans - The if - else statement is two different paths of execution based on whether a condition is True or False.
- If the condition evaluates to True, Python executes the code inside the if block.
- If the condition evaluates to False, Python skips the if block and executes the code inside the else block.
- else condition run when the 1st condition (if) is not satified 
- Only one block (if or else) executes at a time.

Syntax --

if condition:

          # code for True condition

else:

           # code for False condition

Step 1: Check the condition

          |
          v

      Condition
      /       \

    |           |
    v           v

```python
age = int(input("Enter your age = "))

if (age >= 18):
    print("You are eligible for vote")
else:
    print("You are not eligible for vote")
```

```python
num = int(input("Enter your desire no here = "))

if (num >= 0):
    print("Positive No")
else :
    print("Negative No")
```
