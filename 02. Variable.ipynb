### **Q - Variable**
Ans - Variable is just like a container to store your data values.
- **It Changes** - The value can vary whenever you want. (Just like you can swap rice for pulses in a kitchen jar, you can change the value of a 
                   variable anytime in your code.)
- **How it works** - If you store a name or a number in a "block" and give it a name, you can just call that name whenever you need the value back.
- In variable you just assign the value , its not specify what type of value store( int, str) python figures it out on its own.

```python
name = "Sumit"
print(name)
```

### **Q - Identifiers**
Ans - The name you give your variable is called an Identifier.

**There are some basic rules to follow**
- You can use Letters( A-Z or a-z), numbers (0-9), and underscores ( _ ).
- You must start with a letter or an underscore. You cannot start with a number.

Q - Swap two variables a = 5 and b = 10 using a temporary variable temp.

```python
# 1st method
a = 5
b = 10
temp = a
a = b
b = temp
print("a=",a)
print("b=",b)
```

```python
# 2nd method
a = 5
b = 10

a,b = b,a
print("a=",a)
print("b=",b)
```
